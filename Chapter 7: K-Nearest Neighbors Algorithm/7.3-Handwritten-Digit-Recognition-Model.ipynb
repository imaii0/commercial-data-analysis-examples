# 7.3案例实战 - 手写数字识别模型

```python
import pandas as pd
df = pd.read_excel(r"D:\1Ftostudy\商业数据分析\@Python大数据分析与机器学习商业案例实战\@Python大数据分析与机器学习商业案例实战\第7章 K近邻算法\源代码汇总_Jupyter Notebook格式\手写字体识别.xlsx")
df.head()
```

```python
X = df.drop(columns='对应数字')  # 特征变量
y = df['对应数字'] #目标变量
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123)
```

## 决策树模型

```python
# 模型训练及搭建
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=123)
dt.fit(X_train, y_train)
```

```python
# 模型预测及评估
from sklearn.metrics import accuracy_score 
y_pred_dt = dt.predict(X_test)
score_dt = accuracy_score(y_pred_dt, y_test)
print("决策树模型准确率:", score_dt)
print("决策树模型自带score:", dt.score(X_test, y_test))
```

## 随机森林模型

```python
# 模型训练及搭建
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, random_state=123)
rf.fit(X_train, y_train)
```

```python
# 模型预测及评估
y_pred_rf = rf.predict(X_test)
score_rf = accuracy_score(y_pred_rf, y_test)
print("随机森林模型准确率:", score_rf)
print("随机森林模型自带score:", rf.score(X_test, y_test))
```

## 支持向量机

```python
# 模型训练及搭建
from sklearn.svm import SVC
svm = SVC(random_state=123)
svm.fit(X_train, y_train)
```

```python
# 模型预测及评估
y_pred_svm = svm.predict(X_test)
score_svm = accuracy_score(y_pred_svm, y_test)
print("支持向量机模型准确率:", score_svm)
print("支持向量机模型自带score:", svm.score(X_test, y_test))
```
