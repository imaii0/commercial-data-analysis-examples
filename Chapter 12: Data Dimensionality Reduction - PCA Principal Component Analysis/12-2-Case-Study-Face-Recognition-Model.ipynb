```python
import numpy as np
from PIL import Image
import os

# 定义图片文件夹路径
folder_path = r'D:\1Ftostudy\商业数据分析\@Python大数据分析与机器学习商业案例实战\@Python大数据分析与机器学习商业案例实战\第12章 数据降维之PCA主成分分析\源代码汇总_Jupyter Notebook格式（推荐）\olivettifaces'

# 构造所有图片的特征变量
X = [] # 特征变量
for i in names:
    # 使用完整的文件夹路径和os.path.join来拼接路径
    img_path = os.path.join(folder_path, i)
    img = Image.open(img_path)
    img = img.convert('L')
    img = img.resize((32, 32))
    arr = np.array(img)
    X.append(arr.reshape(1, -1).flatten().tolist()) # X为所有人脸图片的特征变量
    print(i) # 查看每次运行结果
```

```python
import pandas as pd
X = pd.DataFrame(X)

X  # 查看400张图片转换后的结果
```

```python
print(X.shape)  # 查看此时的表格结构
```

```python
# 获取目标变量y：第一张图片演示
print(int(names[0].split('_')[0]))
```

```python
folder_path = r'D:\1Ftostudy\商业数据分析\@Python大数据分析与机器学习商业案例实战\@Python大数据分析与机器学习商业案例实战\第12章 数据降维之PCA主成分分析\源代码汇总_Jupyter Notebook格式（推荐）\olivettifaces'

# 批量获取所有图片的目标变量
y = [] # 目标变量
for i in names:
    # 使用os.path.join来正确拼接路径，或者手动添加路径分隔符
    img_path = os.path.join(folder_path, i)
    # 或者：img_path = folder_path + '\\' + i
    img = Image.open(img_path)
    y.append(int(i.split('_')[0]))
print(y) # 查看目标变量,也就是对应的人员编号
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
# 数据降维模型训练
from sklearn.decomposition import PCA
pca = PCA(n_components=100)
pca.fit(X_train)
```

```python
# 对训练集和测试集进行数据降维
X_train_pca = pca.transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
# 我们通过如下代码验证PCA是否降维：
print(X_train_pca.shape)
print(X_test_pca.shape)
```

```python
# 如果想查看此时降维后的X_train_pca和X_test_pca，可以直接将它们打印出来查看，也可以将它们转为DataFrame格式进行查看，代码如下：
print(pd.DataFrame(X_train_pca).head())
print(pd.DataFrame(X_test_pca).head())
```

## 支持向量机(SVM)模型

```python
#模型搭建
from sklearn.svm import SVC
svm_model = SVC(random_state=1)  # 建立SVM模型
svm_model.fit(X_train_pca, y_train)  # 用降维后的训练集进行训练模型
```

```python
#模型预测
y_pred_svm = svm_model.predict(X_test_pca)  # 用降维后的测试集进行测试

# 查看预测结果
print("SVM预测结果:", y_pred_svm)

# 预测准确度
svm_score_pca = svm_model.score(X_test_pca, y_test)
print("SVM降维后准确度:", svm_score_pca)

# 预测值与实际值对比
svm_comparison = pd.DataFrame()
svm_comparison['预测值'] = list(y_pred_svm)
svm_comparison['实际值'] = list(y_test)
print("SVM预测对比:")
print(svm_comparison.head())
```

```python
#模型对比（数据降维与不降维）
svm_model_no_pca = SVC(random_state=1)  # 建立不使用降维的SVM模型
svm_model_no_pca.fit(X_train, y_train)  # 不使用数据降维，直接训练

# 不使用降维的预测准确度
svm_score_no_pca = svm_model_no_pca.score(X_test, y_test)
print("SVM不降维准确度:", svm_score_no_pca)
print("SVM降维提升效果:", svm_score_pca - svm_score_no_pca)
```

## 随机森林(Random Forest)模型

```python
#模型搭建
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(random_state=1)  # 建立随机森林模型
rf_model.fit(X_train_pca, y_train)  # 用降维后的训练集进行训练模型
```

```python
#模型预测
y_pred_rf = rf_model.predict(X_test_pca)  # 用降维后的测试集进行测试

# 查看预测结果
print("随机森林预测结果:", y_pred_rf)

# 预测准确度
rf_score_pca = rf_model.score(X_test_pca, y_test)
print("随机森林降维后准确度:", rf_score_pca)

# 预测值与实际值对比
rf_comparison = pd.DataFrame()
rf_comparison['预测值'] = list(y_pred_rf)
rf_comparison['实际值'] = list(y_test)
print("随机森林预测对比:")
print(rf_comparison.head())
```

```python
#模型对比（数据降维与不降维）
rf_model_no_pca = RandomForestClassifier(random_state=1)  # 建立不使用降维的随机森林模型
rf_model_no_pca.fit(X_train, y_train)  # 不使用数据降维，直接训练

# 不使用降维的预测准确度
rf_score_no_pca = rf_model_no_pca.score(X_test, y_test)
print("随机森林不降维准确度:", rf_score_no_pca)
print("随机森林降维提升效果:", rf_score_pca - rf_score_no_pca)
```

## 逻辑回归(Logistic Regression)模型

```python
#模型搭建
from sklearn.linear_model import LogisticRegression
lr_model = LogisticRegression(random_state=1, max_iter=1000)  # 建立逻辑回归模型
lr_model.fit(X_train_pca, y_train)  # 用降维后的训练集进行训练模型
```

```python
#模型预测
y_pred_lr = lr_model.predict(X_test_pca)  # 用降维后的测试集进行测试

# 查看预测结果
print("逻辑回归预测结果:", y_pred_lr)

# 预测准确度
lr_score_pca = lr_model.score(X_test_pca, y_test)
print("逻辑回归降维后准确度:", lr_score_pca)

# 预测值与实际值对比
lr_comparison = pd.DataFrame()
lr_comparison['预测值'] = list(y_pred_lr)
lr_comparison['实际值'] = list(y_test)
print("逻辑回归预测对比:")
print(lr_comparison.head())
```
