## 通过efficient-apriori库来实现Apriori算法

```python
from efficient_apriori import apriori

# 使用 efficient-apriori 库实现
itemsets, rules = apriori(symptoms, min_support=0.1, min_confidence=0.7)

# 打印频繁项集
print("频繁项集:")
for itemset, support in itemsets.items():
    if itemset >= 1:  # 只显示包含至少1个项的项集
        for key, value in itemsets[itemset].items():
            print(f"{set(key)}: {value}")

# 打印关联规则
print("\n关联规则:")
for rule in rules:
    x = ', '.join(rule.lhs)  # 前件
    y = ', '.join(rule.rhs)  # 后件
    print(f"{x} → {y} (置信度: {rule.confidence:.3f})")
```

## 通过pymining库来实现Apriori算法

```python
from pymining import itemmining, assocrules, perftesting

# 使用 pymining 库实现
relim_input = itemmining.get_relim_input(symptoms)
item_sets = itemmining.relim(relim_input, min_support=0.1)

# 生成关联规则
rules = assocrules.mine_assoc_rules(item_sets, min_support=0.1, min_confidence=0.7)

# 打印结果
print("关联规则:")
for rule in rules:
    antecedent = ', '.join(rule[0])  # 前件
    consequent = ', '.join(rule[1])  # 后件
    support = rule[2]
    confidence = rule[3]
    print(f"{antecedent} → {consequent} (支持度: {support:.3f}, 置信度: {confidence:.3f})")
```

## 不依赖外部库来实现Apriori算法

```python
def generate_candidates(itemset, length):
    """生成候选项集"""
    return set([i.union(j) for i in itemset for j in itemset if len(i.union(j)) == length])

def prune_itemset(itemset, previous_itemset, length):
    """剪枝步骤"""
    candidates = set()
    for item in itemset:
        # 检查所有长度为length-1的子集是否在previous_itemset中
        subsets = [frozenset([x for x in item if x != y]) for y in item]
        if all(subset in previous_itemset for subset in subsets):
            candidates.add(item)
    return candidates

def apriori_pure(symptoms, min_support=0.1):
    """纯Python实现的Apriori算法"""
    from collections import defaultdict
    
    # 计算支持度
    def get_support(itemset):
        count = 0
        for transaction in symptoms:
            if itemset.issubset(transaction):
                count += 1
        return count / len(symptoms)
    
    # 找出频繁1项集
    items = set()
    for transaction in symptoms:
        for item in transaction:
            items.add(frozenset([item]))
    
    frequent_itemsets = {}
    current_itemset = set()
    
    for item in items:
        support = get_support(item)
        if support >= min_support:
            current_itemset.add(item)
    
    k = 1
    frequent_itemsets[k] = current_itemset
    
    while len(current_itemset) > 0:
        k += 1
        # 生成候选项集
        candidates = generate_candidates(current_itemset, k)
        # 剪枝
        candidates = prune_itemset(candidates, current_itemset, k)
        
        current_itemset = set()
        for candidate in candidates:
            support = get_support(candidate)
            if support >= min_support:
                current_itemset.add(candidate)
        
        if len(current_itemset) > 0:
            frequent_itemsets[k] = current_itemset
        else:
            break
    
    return frequent_itemsets

# 使用纯Python实现
frequent_itemsets = apriori_pure(symptoms, min_support=0.1)

print("频繁项集:")
for k, itemsets in frequent_itemsets.items():
    for itemset in itemsets:
        itemset_str = ', '.join(itemset)
        support = sum(1 for transaction in symptoms if itemset.issubset(transaction)) / len(symptoms)
        print(f"{itemset_str}: {support:.3f}")  # 修正了这一行
```
