# 10.6 LightGBM算法案例实战2 - 广告收益回归预测模型

```python
# 读取数据
import pandas as pd
df = pd.read_excel(r"D:\1Ftostudy\商业数据分析\@Python大数据分析与机器学习商业案例实战\@Python大数据分析与机器学习商业案例实战\第10章 机器学习神器：XGBoost&LightGBM模型\源代码汇总_Jupyer Notebook\广告收益数据.xlsx")
df.head()
```

```python
# 1.提取特征变量和目标变量
X = df.drop(columns='收益') 
y = df['收益']

# 2.划分训练集和测试集
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123)
```

## 线性回归模型

```python
# 模型搭建
from sklearn.linear_model import LinearRegression

model_lr = LinearRegression()
model_lr.fit(X_train, y_train)
```

```python
# 模型预测及评估
y_pred_lr = model_lr.predict(X_test)

# 评估
from sklearn.metrics import r2_score, mean_squared_error

print("线性回归 R²:", r2_score(y_test, y_pred_lr))
print("线性回归 MSE:", mean_squared_error(y_test, y_pred_lr))

# 预测值与实际值对比
df_compare_lr = pd.DataFrame({'预测值': y_pred_lr, '实际值': y_test})
print(df_compare_lr.head())
```

## 随机森林回归

```python
# 模型搭建
from sklearn.ensemble import RandomForestRegressor

model_rf = RandomForestRegressor(n_estimators=100, random_state=123)
model_rf.fit(X_train, y_train)
```

```python
# 模型预测及评估
y_pred_rf = model_rf.predict(X_test)

# 评估
print("随机森林 R²:", r2_score(y_test, y_pred_rf))
print("随机森林 MSE:", mean_squared_error(y_test, y_pred_rf))

# 预测值与实际值对比
df_compare_rf = pd.DataFrame({'预测值': y_pred_rf, '实际值': y_test})
print(df_compare_rf.head())

# 特征重要性
print("特征重要性:", model_rf.feature_importances_)
```

## 支持向量机回归模型

```python
# 模型搭建
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler

# SVR 对特征尺度敏感，建议标准化
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model_svr = SVR(kernel='rbf', C=1.0, epsilon=0.1)
model_svr.fit(X_train_scaled, y_train)
```

```python
# 模型预测及评估
y_pred_svr = model_svr.predict(X_test_scaled)

# 评估
print("SVR R²:", r2_score(y_test, y_pred_svr))
print("SVR MSE:", mean_squared_error(y_test, y_pred_svr))

# 预测值与实际值对比
df_compare_svr = pd.DataFrame({'预测值': y_pred_svr, '实际值': y_test})
print(df_compare_svr.head())
```
